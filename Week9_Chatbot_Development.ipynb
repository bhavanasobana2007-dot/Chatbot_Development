```python

import os
import csv
import pandas as pd
import ollama
from datetime import datetime

print("All libraries imported successfully!")
```

```python

# Find the project directory
current_dir = os.getcwd()

if os.path.basename(current_dir).lower() == "notebook":
    project_dir = os.path.dirname(current_dir)
else:
    project_dir = current_dir

dataset_dir = os.path.join(project_dir, "dataset")
os.makedirs(dataset_dir, exist_ok=True)

questions_file = os.path.join(dataset_dir, "sample_questions.csv")
history_file = os.path.join(dataset_dir, "chat_history.csv")
responses_file = os.path.join(dataset_dir, "responses.csv")

print("Project folder:", project_dir)
print("Dataset folder:", dataset_dir)
```

```python

sample_questions = [
    "What is Python?",
    "What is Artificial Intelligence?",
    "What is Machine Learning?",
    "What is a Large Language Model?",
    "What is a database?",
    "What is DBMS?",
    "What is SQL?",
    "What is a primary key?",
    "What is a foreign key?",
    "What is normalization in DBMS?",
    "What is a Python variable?",
    "What is a Python function?",
    "What is a loop in Python?",
    "What is a list in Python?",
    "What is an algorithm?",
    "What is a data structure?",
    "What is cloud computing?",
    "What is cybersecurity?",
    "Explain the difference between AI and ML.",
    "Write a Python program to add two numbers."
]

questions_df = pd.DataFrame(sample_questions, columns=["question"])
questions_df.to_csv(questions_file, index=False)

print("Sample questions created successfully!")
print("Total questions:", len(questions_df))
display(questions_df)
```

```python

MODEL = "llama3.2:1b"

test_question = "What is Python? Explain in simple English."

response = ollama.chat(
    model=MODEL,
    messages=[
        {
            "role": "system",
            "content": (
                "You are an education assistant. "
                "Answer in simple English with clear examples. "
                "If you are unsure, say so."
            )
        },
        {"role": "user", "content": test_question}
    ]
)

print("Question:", test_question)
print("Chatbot Response:", response["message"]["content"])
```

```python

import pandas as pd
import os

questions = [
    "What is Python?",
    "What is Artificial Intelligence?",
    "What is Machine Learning?",
    "What is a Large Language Model?",
    "What is a database?",
    "What is DBMS?",
    "What is SQL?",
    "What is a primary key?",
    "What is a foreign key?",
    "What is normalization in DBMS?",
    "What is a Python variable?",
    "What is a Python function?",
    "What is a loop in Python?",
    "What is a list in Python?",
    "What is an algorithm?",
    "What is a data structure?",
    "What is cloud computing?",
    "What is cybersecurity?",
    "Explain the difference between AI and ML.",
    "Write a Python program to add two numbers."
]

questions_df = pd.DataFrame(questions, columns=["question"])

print("Total questions:", len(questions_df))
display(questions_df)
```

```python

results = []

for question in questions_df["question"].dropna():
    try:
        response = ollama.chat(
            model=MODEL,
            messages=[
                {
                    "role": "system",
                    "content": (
                        "You are an education assistant. "
                        "Answer in simple English. "
                        "Give correct and relevant answers. "
                        "If you are unsure, say so."
                    )
                },
                {"role": "user", "content": question}
            ]
        )

        answer = response["message"]["content"]
        status = "Success"

    except Exception as error:
        answer = str(error)
        status = "Error"

    results.append({
        "timestamp": datetime.now().isoformat(timespec="seconds"),
        "question": question,
        "response": answer,
        "status": status
    })

    print("Question:", question)
    print("Response:", answer)
    print("-" * 50)

responses_df = pd.DataFrame(results)
responses_df.to_csv(responses_file, index=False)

print("Testing completed!")
print("Total tested:", len(responses_df))
print("Successful responses:", (responses_df["status"] == "Success").sum())
print("Saved to:", responses_file)
```

```python

import pandas as pd
import os

if results:
    responses_df = pd.DataFrame(results)

    output_path = os.path.join(dataset_dir, "responses.csv")
    responses_df.to_csv(output_path, index=False)

    print("Results saved successfully!")
    print("Total records:", len(responses_df))
    print("Saved location:", output_path)

    display(responses_df.head())
else:
    print("No results found. Run the testing cell first.")
```
